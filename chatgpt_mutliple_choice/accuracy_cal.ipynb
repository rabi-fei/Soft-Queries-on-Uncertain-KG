```python
import os
import json
import pickle
import random
import numpy as np
```

```python
path_profix = "oct8"
```

```python
pickle_files = {}
for t in ["type0000",
          "type0001",
          "type0002",
          "type0004",
          "type0005",
          "type0008",
          "type0009",
          "type0007"]:
    path_to_pickle = path_profix + "/" + t + ".pickle"
    if os.path.isfile(path_to_pickle):
        if t not in pickle_files:
            pickle_files[t] = [path_to_pickle]
        else:
            pickle_files[t].append(path_to_pickle)
pickle_files
```

```python
pickle_output_files = {}
for t in ["type0000",
          "type0001",
          "type0002",
          "type0004",
          "type0005",
          "type0008",
          "type0009",
          "type0007"]:
    path_to_pickle = path_profix + "/" + t + "_outputs.pickle"
    if os.path.isfile(path_to_pickle):
        if t not in pickle_output_files:
            pickle_output_files[t] = [path_to_pickle]
        else:
            pickle_output_files[t].append(path_to_pickle)
pickle_output_files
```

```python
overall_accuracy = {"correct":0,"total":0}
for t in ["type0000",
          "type0001",
          "type0002",
          "type0004",
          "type0005",
          "type0008",
          "type0009"]:
    dataset = pickle.load(open(pickle_files[t][0],"rb"))
    output = pickle.load(open(pickle_output_files[t][0],"rb"))
    
    total = len(output)
    correct = 0
    for i in range(total):
        if dataset[i][0] == output[i][0] and dataset[i][5] == output[i][1]:
            correct = correct + 1
        elif dataset[i][0] != output[i][0]:
            print("Index Error!")
    
    overall_accuracy["correct"] = overall_accuracy["correct"] + correct
    overall_accuracy["total"] = overall_accuracy["total"] + total
    
    print(t, "accuracy:", format(correct / total, '.2f'), "total:", total)

print("overall", "accuracy:", format(overall_accuracy["correct"] / overall_accuracy["total"], '.2f'), "total:", overall_accuracy["total"])
```
