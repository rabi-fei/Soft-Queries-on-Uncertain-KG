# figure out the tune learning rate and sigma over small model

```python
def command(batch_size, learning_rate, sigma, neg_sigma_scaling):
    commandline = ("cd /mount/kgtvr && python3 task_dependent_train.py "
                   "--embedding_dim=100 "
                   "--device=cuda:0 "
                   "--checkpoint_path=pretrain/FB15k-237-model-rank-100-epoch-100-1602503352.pt "
                   "--objective noisy "
                   "--output_dir /results "
                   f"--neg_sigma_scaling={neg_sigma_scaling} "
                   f"--learning_rate={learning_rate} "
                   f"--batch_size={batch_size} "
                   f"--sigma={sigma} ")
    command = ('ngc batch run '
               f'--name "ml-model.TVR.bs{batch_size}_lr{learning_rate}_sgm{sigma}_nss{neg_sigma_scaling}" '
               '--preempt RUNONCE '
               '--ace nv-us-west-2 '
               '--instance dgx1v.16g.1.norm '
               '--result /results '
               '--image "nvidia/pytorch:22.04-py3" '
               '--org nvidian '
               '--team sae '
               f'--commandline "{commandline}" '
               '--workspace 8sr-vg7_STK0qaux2KusFA:/mount/kgtvr:RW')
    return command
```

```python
batch_size_list = [256, 2048]
learning_rate_list = [1e-3, 1e-4]
sigma_list = [1, 3]
neg_sigma_scaling = [1, 10]

for bs, lr, sigma, nss in product(batch_size_list, learning_rate_list, sigma_list, neg_sigma_scaling):
    cmd = command(bs, lr, sigma, nss)
    print(cmd)
    os.system(cmd)
```
